```python
from PIL import Image
from PIL import ImageOps
import numpy as np
import pandas as pd
import sys

#from gdpc import __url__, Editor, Block

#editor = Editor(buffering=True)
#buildArea = editor.getBuildArea()
#editor.loadWorldSlice(cache=True)
#heightmap = editor.worldSlice.heightmaps["MOTION_BLOCKING_NO_LEAVES"]

# Open image
img = Image.open("maps/detailed_bedrock_geology.png")
img = img.convert('RGB')

# Resize smoothly down
img_r = img.resize((577, 428), resample=Image.Resampling.NEAREST)

#create numpy array from resized map
arr = np.array(img_r)
arr = arr[:,:,:3]
```

```python
#combine R, G, and B values into single rgb string
rows, cols, z = arr.shape
arr_rgb = np.empty(shape=(rows,cols), dtype = int)
for x in range(rows):  
    for y in range(cols):
        arr_rgb[x,y] = ''.join(str(n) for n in arr[x,y])
```

```python
geo = open("detailed_geology_d.txt", "r")
lst=[]
for line in geo:
    lst.append([ (x) for x in line.split()])

rgb = [ x[0] for x in lst]
block = [ x[1] for x in lst]
rgb = [int(s) for s in rgb]
type(block)
```

```python
#remove unwanted RGB values
rows, cols= arr_rgb.shape
arr_fix = np.empty(shape=(rows,cols), dtype = int)
for x in range(rows):
    for y in range(cols):
        for n in range(0,1033):
            if (np.array_equal(arr_rgb[x,y],rgb[n])
            or np.array_equal(arr_rgb[x,y],[255255255])):
                arr_fix[x,y] = arr_rgb[x,y]
            else:
                arr_fix[x,y] = arr_fix[x-1,y]
```

```python
rows, cols= arr_rgb.shape
arr_fix = np.zeros(shape=(rows,cols), dtype = int)
for n in range(1034):
    for x in range(rows):
        for y in range(cols):
            if arr_rgb[x,y] == rgb[n]:
                arr_fix[x,y] = arr_rgb[x,y]
```

```python
for x in range(rows):
    for y in range(cols):
        if arr_fix[x,y] == 0:
            arr_fix[x,y] = arr_fix[x-1,y]
```
